# 4. RAG Pipeline Evaluation
This notebook evaluates the RAG system against the Insurellm benchmark dataset (`evaluation/tests.jsonl`).

```python
import os
from dotenv import load_dotenv
from evaluation import test, eval
from implementation.answer import answer_question, fetch_context

load_dotenv(override=True)
```

```python
# 1. Load test cases
tests = test.load_tests()
print(f"Total test cases loaded: {len(tests)}")
```

```python
# 2. Preview sample test cases
for t in tests[:3]:
    print("Category   :", t.category)
    print("Question   :", t.question)
    print("Keywords   :", t.keywords)
    print("Reference  :", t.reference_answer)
    print("-" * 60)
```

```python
# 3. Test a single evaluation
sample_test = tests[0]
answer, docs = answer_question(sample_test.question)
print("Question :", sample_test.question)
print("\nGenerated Answer :", answer)

res = eval.evaluate_test(sample_test, answer)
print("\nEvaluation Result :", res)
```

```python
# 4. Run evaluation benchmark (e.g. first 5 tests)
summary = eval.run_evaluation(
    tests=tests,
    answer_fn=lambda q: answer_question(q)[0],
    limit=5,
    verbose=True
)
```

```python
# 5. Display evaluation summary metrics
print("=" * 50)
print("             EVALUATION SUMMARY")
print("=" * 50)
print(f"Total Tests Evaluated : {summary['total_tests']}")
print(f"Tests Passed Fully    : {summary['passed_count']}")
print(f"Strict Pass Rate      : {summary['accuracy'] * 100:.1f}%")
print(f"Average Keyword Match : {summary['average_score'] * 100:.1f}%")
print("\nCategory Breakdown:")
for cat, score in summary['category_performance'].items():
    print(f"  • {cat:15s}: {score * 100:.1f}%")
print("=" * 50)
```
